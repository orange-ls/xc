#### **归一化：把不同量纲、不同取值范围的数据“拉齐”到同一尺度，避免模型“偏心”某些特征。**

| 方法名称           | 公式                                              | 特点                  |
| -------------- | ----------------------------------------------- | ------------------- |
| **Min-Max归一化** | $x' = \frac{x - x_{\min}}{x_{\max} - x_{\min}}$ | 缩放到 \[0, 1]，受异常值影响大 |
| **Z-score标准化** | $x' = \frac{x - \mu}{\sigma}$                   | 均值为0，方差为1，适用于正态分布   |
| **Robust标准化**  | $x' = \frac{x - \text{median}}{\text{IQR}}$     | 抗异常值能力强             |


```python
import numpy as np
from sklearn import datasets
```

```python
# 加载鸢尾花数据集
iris = datasets.load_iris()
```

```python
X = iris.data
y = iris.target
```

### **最大最小值归一化**

```python
# 先打印出前五行 看看数据
X[:5]
```

```python
# 找出第一列最大值和最小值
np.max(X[:,0]),np.min(X[:,0])
```

```python
# 使用公式，计算出每个缩放后的值
X[:,0] = (X[:,0] - np.min(X[:,0])) / (np.max(X[:,0]) - np.min(X[:,0]))
```

```python
X[:5,0]
```

```python
# 同理，处理后面几列数据
X[:,0] = (X[:,0] - np.min(X[:,0])) / (np.max(X[:,0]) - np.min(X[:,0]))
X[:,0] = (X[:,0] - np.min(X[:,0])) / (np.max(X[:,0]) - np.min(X[:,0]))
X[:,0] = (X[:,0] - np.min(X[:,0])) / (np.max(X[:,0]) - np.min(X[:,0]))
```
