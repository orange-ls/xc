```python
import numpy as np
```

```python
X = np.arange(1,21).reshape(4,5)
X
```

```python
q = np.linspace(0,100,5)
q
```

#### np.percentile(a, q, axis, method) 
#### 百分位数（Percentile）：表示在一个数据集中，有百分之多少的数据点小于或等于这个值。

```python
np.percentile(X,q)
```

#### np.cumsum(X,axis=1)    axis=1 - 按列累加；axis=0 - 按行累加

```python
np.cumsum(X,axis=0)
```

```python
np.cumsum(X,axis=1)
```

#### np.prod() 乘积

```python
np.prod(X,axis=0)
```
