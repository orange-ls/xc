```python
import numpy as np
```

```python
A = np.arange(16).reshape(4,4)
A
```

```python
a = np.array([6,6,6,6])
a
```

#### np.concatenate((array1,array2),axis) 拼接两个数组，需要相同维度

```python
X = np.concatenate((A,a.reshape(1,4)))
print(X,"\n")
X_1 = np.concatenate((A,a.reshape(4,1)),axis=1)
print(X_1)
```

#### 另外一种拼接方式 (不用统一数组的维度,但是还是要确保拼接方向的元素一致)：
#### 1、竖直方向的合并：np.vstack((array1,array2)) 
#### 2、水平方向的合并: np.hstack((array1,array2)) 

```python
X1 = np.vstack((A,a))
X1
```

```python
X2 = np.hstack((A,a.reshape(4,1)))
X2
# a.reshape(4,1)
```

```python
b = np.arange(6).reshape(2,3)
b.T
```

#### 数组拆分

#### np.split(array,[],axis) axis默认为0，按行拆分

```python
np.split(X1,[-1])
```

```python
R1,R2 = np.split(X2,[-1],axis=1)
print(R1)
print(R2)
```

#### vsplit、hsplit

```python
np.vsplit(X1,[2])
```

```python
np.hsplit(X1,[1,3])
```
