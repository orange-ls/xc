```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

```python
# 1. 生成 4 簇，每簇标准差不一样
a, b = make_blobs(n_samples=800,
                  centers=[[0,0], [6,6], [-6,6], [0,-10]],
                  cluster_std=[0.8, 1.5, 0.5, 2.0],
                  random_state=111)
a, b
```

```python
# 获取唯一的标签
unique_labels = set(b)

# 为每个标签选择颜色，这里我们使用一个颜色列表，确保有4种颜色
colors = ['red', 'blue', 'green', 'orange']

for i, label in enumerate(unique_labels):
    # 选择属于当前标签的点
    idx = (b == label)
    plt.scatter(a[idx, 0], a[idx, 1], c=colors[i], label=f'Cluster {label}', s=25)

plt.legend()
plt.title('make_blobs with legend')
plt.show()
```
