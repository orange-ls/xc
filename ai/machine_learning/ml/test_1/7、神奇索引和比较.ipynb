```python
import numpy as np
```

```python
# x = np.random.randint(1,10,10)
x = np.arange(10)
x
```

```python
# 从索引为2一直取到6(不包括6)，步长为1
x[2:6:1]
```

```python
# 使用神奇索引
index = [2,4,6,8]
x[index]
```

```python
# 索引是二维的，结果也会是二维的
# 传入的值可以是数组，也可以不是
index = [[1,3,5],[2,4,6]]
index_r = np.array([[1,3,5],[2,4,6]])
x[index],x[index_r]
```

### 神奇索引用于二维数组

```python
X = x.reshape(2,-1)
X
```

```python
# 取出 (0,0)、(1,2)、(0,4)
row = np.array([0,1,0])
col = np.array([0,2,4])
X[row,col]
```

```python
# 取出 (0,2)、(0,4)、(0,1)
X[0,[2,4,1]]
```

### 比较
### 比较结果和神奇索引的使用

```python
x
```

```python
# x>5 返回出True或False
x>5,x[x>5]
```

```python
x[x%2==0]
```

### 比较的更多应用

```python
# count_nonzero()返回数组中不是0的元素个数，但是，x<5返回的是bool类型，所以包括0
np.count_nonzero(x),np.count_nonzero(x<5)
```

```python
# any()函数
np.any(x>5)
```

```python
# all()
np.all(x>0)
```

```python
# sum()
np.sum(x%2==0)
```

```python
X = x.reshape(2,-1)
# axis=几，就把第几维“压扁”，结果就是剩余的维数
np.sum(X%2==0,axis=1),np.sum(X%2==0,axis=0)
# 例：原维度是(2,5)，axis=1时，1维被压缩去掉，剩余0维(2,)
```

### 与& 或| 非~

```python
x[(x>3)&(x<8)]
```

```python
np.all((x<2)|(x>8)),np.any((x<2)|(x>8))
```
