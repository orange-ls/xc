```python
import numpy as np
```

```python
# x = np.random.randint(1,10,10)
x = np.arange(10)
x
```

```python
# 从索引为2一直取到6(不包括6)，步长为1
x[2:6:1]
```

```python
# 使用神奇索引
index = [2,4,6,8]
x[index]
```

```python
# 索引是二维的，结果也会是二维的
# 传入的值可以是数组，也可以不是
index = [[1,3,5],[2,4,6]]
index_r = np.array([[1,3,5],[2,4,6]])
x[index],x[index_r]
```
