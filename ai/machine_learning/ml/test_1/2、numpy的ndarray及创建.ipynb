## 创建数组的特殊函数

#### ones(shape, dtype=None, order='C') 创建元素全为1 的数组，可用.reshape()来改变维度

```python
d = np.ones(9)
print(f"{d}\n")
e = np.ones((3,3))
print(e,'\n')
f = np.ones_like(e)
print(f)
```

#### zeros(shape, dtype=None, order='C') 创建全0的数组,用法与ones相同

```python
np.zeros(3)
```

#### full(shape, fill_value, dtype=None, order='C') 创建全为设定值的数组

```python
np.full((2,3),666)
```

#### empty(shape, dtype=float, order='C') 创建未定义 数组

```python
np.empty((2,3))
```

#### arange([start, ]stop, [step, ]dtype=None) 创建均匀间隔数组，默认从0开始、步长为1

```python
np.arange(1,10,2)
```

#### linspace(start, stop[, num, endpoint=True, retstep=False, dtype=None, axis=0])
#### 在指定的区间内，创建包含 *指定数量* 均匀间隔数值的一维数组

```python
np.linspace(1,10,5)
```

### random模块

random.random()

```python
# 没有参数时，会生成一个[0,1)的随机数
np.random.random()
```

```python
np.random.random(5)
```

```python
#  入参为元组时，可以生成指定维度数组
np.random.random((2,3))
```

```python
np.random.random((2,3,4))
```

```python
a = np.random.random((2,3,2))
a.shape
```

#### random.rand()

```python
# 和random.random()不同 rand()的入参不用是元组也可以生成对应的数组
np.random.rand(2,3,4)
```

#### random.randint() 生成随机整数，入参 [ , )

```python
np.random.randint(2,6,size=10)
```

```python
np.random.randint(2,6,size=(3,2))
```

#### random.seed() 随机种子

```python
np.random.seed(123)
np.random.rand(2,3,4)
```

#### np.random.randn() 返回标准正态分布的随机数，均值为0，标准差为1。用法于rand()类似

```python
np.random.randn(2,3)
```

#### np.random.normal( , , ) 可指定均值和标准差，用法与randint()类似 

```python
np.random.normal(10,100,size=(2,3))
```

#### random.uniform() 返回均匀分布的随机数据。用法与randn()类似

```python
help(np.random.uniform)
```

```python
import matplotlib.pyplot as plt
s = np.random.uniform(-1,0,1000)
# print(s)
count, bins, ignored = plt.hist(s, 15, density=True)
plt.plot(bins, np.ones_like(bins), linewidth=2, color='r')
plt.show()
```
