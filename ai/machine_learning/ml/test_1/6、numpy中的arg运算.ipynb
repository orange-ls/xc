```python
import numpy as np
```

```python
a = np.random.normal(0,1,size=1000000)
a
```

```python
# 均值、方差
np.mean(a),np.std(a)
```

```python
np.min(a)
```

```python
# argmin()可以找出最小元素的索引，根据索引找到的最小值与min()结果一样
print(np.argmin(a))
print(a[np.argmin(a)])
```

```python
# argmax()同理
print(np.argmax(a))
print(a[np.argmax(a)])
```

### 排序

```python
c = np.arange(20)
c
```

```python
# 使用random.shuffle,会打乱原数组的顺序，并保存在原变量中
np.random.shuffle(c)
c
```

```python
# 使用np.sort(),重新排序，顺序，不会改变原变量中的顺序
np.sort(c), c
```

```python
# 使用python原生sort() 可以修改原变量
d = c
d.sort()
d
```

```python
# 使用切片来实现倒序 d[::-1]
d[::-1]
```

### 使用索引

```python
np.random.shuffle(c)
c
```

```python
# 返回正序元素的索引
np.argsort(c)
```

### partition分区 将数组部分排序，使得第k小的元素在正确的位置上
### np.partition(a, kth, axis=-1, kind='introselect', order=None)

```python
np.partition(c,5)
```

```python
np.partition(c,[5,10])
```

```python
# np.argpartition(),返回索引
np.argpartition(c,5)
```

### 高维的情况

```python
X = np.random.randint(10,size=(5,5))
X
```

```python
# np.sort() 默认以最高维度排序。二维axis=1,按列；三维axis=2
np.sort(X)
```

```python
np.sort(X,axis=0)
```
