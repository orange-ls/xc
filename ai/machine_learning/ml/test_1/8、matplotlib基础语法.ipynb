```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-10,10,100)
y = x*x
```

```python
plt.plot(x,y)
plt.show()
```

```python
# 如果点的个数少，图像就会变成折线
x = np.linspace(-10,10,5)
y = x*x
plt.plot(x,y)
plt.show()
```

```python
x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x,y1)
plt.plot(x,y2)
plt.show()
```

### 设置曲线的颜色 plt.plot(x, y, color='green') 其他颜色：'red', 'blue', 'green'...
### 设置曲线的线形：默认为实线，虚线:"--"、点线:":"、点划线:"-."

```python
x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x,y1,color='green',linestyle='-.')
plt.plot(x,y2,color='red',linestyle=':')
plt.show()
```

```python
plt.plot(x,y1,color='green',linestyle=':')
plt.plot(x,y2)
# 设置范围
plt.xlim(-5,15)
plt.ylim(-2,2)
plt.show()
```

```python
plt.plot(x,y1,color='red',linestyle=':')
plt.plot(x,y2)
# 使用axis设置范围
plt.axis([-5,15,-2,2])
plt.show()
```

```python
plt.plot(x,y1,color='red',linestyle=':')
plt.plot(x,y2)
# 设置x,y轴的名字
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

```python
# 设置两条线的名字
plt.plot(x,y1,color='red',linestyle=':',label="sinx")
plt.plot(x,y2,label="cosx")
# 设置两条线的名字后，还需要增加图例才能显示区线的名字
plt.legend()
# 给图标增加一个标题
plt.title("This is my first plot")
plt.show()
```

### 离散情况下，不使用plt.plot，使用plt.scatter()

```python
plt.scatter(x,y1)
plt.show()
```

```python
x = np.random.normal(0,1,100000)
y = np.random.normal(0,1,100000)
plt.scatter(x,y,color="yellow")
plt.show()
```

```python
# 不透明参数alpha，[0，1] 0为透明，1为不透明
x = np.random.normal(0,1,100000)
y = np.random.normal(0,1,100000)
plt.scatter(x,y,alpha=0.1)
plt.show()
```
