```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-10,10,100)
y = x*x
```

```python
plt.plot(x,y)
plt.show()
```

```python
# 如果点的个数少，图像就会变成折线
x = np.linspace(-10,10,5)
y = x*x
plt.plot(x,y)
plt.show()
```

```python
x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x,y1)
plt.plot(x,y2)
plt.show()
```

### 设置曲线的颜色 plt.plot(x, y, color='green') 其他颜色：'red', 'blue', 'green'...
