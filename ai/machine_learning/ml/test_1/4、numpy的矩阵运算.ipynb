### 一元运算（有一个操作数）

```python
X1 = np.arange(1,7,0.5).reshape(3,4)
X1
```

```python
# 绝对值 abs
np.abs(X1)
```

```python
# 开方 sqrt
np.sqrt(X1)
```

```python
# 平方 square
np.square(X1)
```

```python
# 自然常数e的指数运算 exp
np.exp(X1)
```

```python
# 对数运算 log() - 以e为底。 logn() - 以n为底
print(np.log(X1))
print(np.log2(X1))
```

```python
# 向上取整 cell    向下取整 floor
print(np.ceil(X1))
print(np.floor(X1))
```

```python
# 四舍五入函数 round
np.round(X1)
```

```python
# 三角函数 sin、cos、tan
print(np.sin(X1))
print(np.cos(X1))
print(np.tan(X1))
```

### 二元运算（有两个操作数）

#### 最简单的二元运算 就是 加减乘除

```python
# 加法
X1 + 1
# 相当于使用 np.add(X1,1)
```

```python
# 乘、除
print(X1*2)
print(2*X1)
print(X1/2)
```

```python
# 整除、取余、幂运算
print(X1//2)
print(X1%2)
print(X1**2)
```

#### 矩阵运算

```python
# 创建两个矩阵
Y = np.linspace(10,40,4).reshape(2,2)
Y1 = np.linspace(1,4,4).reshape(2,2)
Y,Y1
```

#### 哈达玛积（对应位置的元素相乘，不是矩阵的相乘）
#### 哈达玛积的规则：两个数组必须具有相同的形状，或者满足广播机制的兼容形状。
#### 广播机制：从最右边的维度开始向左对齐。在大小为1或缺失的维度上，通过复制数据来扩展数组，以匹配另一个数组的维度大小。

```python
# 最简单的哈达玛积：标量与数组（array * n）
Y * Y1
```

#### 矩阵相乘（点积） np.dot(array1,arry2) 

```python
# Y.dot(Y1)
np.dot(Y,Y1)
```

```python
# 对应位置相除，同样有广播机制
Y/Y1
```

```python
# 广播机制，Y的每一行都和A点积，并将结果简化为一维输出
A = np.array([1,4])
B = np.dot(Y,A)
A,B
```

```python
# 广播机制，并不是把[1,4]转换成[[1,4],[1,4]]后再点积
A = np.array([[1,4],[1,4]])
np.dot(Y,A)
```

#### 逆矩阵 np.linalg.inv(A)

```python
C = np.linalg.inv(Y)
C
```

```python
np.dot(C,Y)
```

```python
np.dot(Y,C)
```
